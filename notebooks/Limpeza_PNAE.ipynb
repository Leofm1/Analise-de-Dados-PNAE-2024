```python
import pandas as pd

caminho = "../dados/brutos/DadosFinanceirosdoPNAE_RedesEstadual_Distrital_Municipal_PorEntidadeExecutora_2024.xlsx"

df_bruto = pd.read_excel(caminho, sheet_name="Planilha1")

print("Estas são as dimensões e a tabela de dados Financeiros do PNAE 2024 bruta:")
print(df_bruto.shape)
df_bruto.head(10)
```

```python
#Este arquivo tem 4 detalhes que precisam ser resolvidos.


#1 - As primeiras 4 linhas não são dados, então precisamos pular essas linhas.
df = pd.read_excel(caminho, sheet_name="Planilha1", skiprows=4)

#2 - A planilha tem 5 colunas, sendo as subsequentes colunas irrelevantes, então precisamos manter apenas as 5 primeiras colunas. 
df = df.iloc[:, :5]

print(df.shape)
df.head(10)
```

```python
#3 - Temos linhas com o valor "Total" na coluna "UF", seguidas de uma linha em branco
# que não são dados relevantes, então precisamos remover essas linhas.

#limpeza de linhas com "Total" na coluna "UF"
df = df[~df["UF"].astype(str).str.contains("TOTAL",case=False, na=False)].copy()

#limpeza de linhas em branco
df = df.dropna(how='all')

#verificando se ainda existem valores nulos no dataframe
df.isna().sum()
```

```python
#4 - As colunas de dados numericos estão como texto, então precisamos converter essas colunas para o tipo numérico.
def limpar_float(x):
    if isinstance(x, (float)):
        return float(x)
    x_str = str(x).strip()
    return float(x_str.replace('.', '').replace(',', '.'))

def limpar_int(x):
    if isinstance(x, (int)):
        return int(x)
    x_str = str(x).strip()
    return int(x_str.replace('.', ''))

# Em uma instância anterior deste código, as funções anteriores não verificavam se o valor era float ou int, assim, acabavam multiplicando o valor por 100
# As funções, então, foram corrigidas para verificar se o valor já é do tipo correto antes de tentar convertê-lo.

df["VALOR_PREVISTO"] = df["VALOR_PREVISTO"].apply(limpar_float)

df["QTDE_ALUNO"] = df["QTDE_ALUNO"].apply(limpar_int)

df.head(10)
```

```python
#5 - Traduzindo os códigos para nomes legíveis
mapa_acoes = {
    'PNACN': 'Creche',
    'PNAPN': 'Pré-escola',
    'PNAFN': 'Ensino Fundamental',
    'PNAEN': 'EJA (Educação de Jovens e Adultos)',
    'PNAEE': 'Educação Especial (AEE)',
    'PNAMN': 'Ensino Médio',
    'PNAMI': 'Ensino Médio em Tempo Integral (EMTI)',
    'PNAIN': 'Indígena',
    'PNAQN': 'Quilombola'
}

df['MODALIDADE'] = df['ACAO'].map(mapa_acoes)

#Removendo a coluna ACAO, pois agora temos a coluna MODALIDADE com os nomes legíveis.
df = df.drop(columns=['ACAO'])

df.head(10)
```

```python
#Por fim, salvando o dataframe limpo em um novo arquivo csv.
caminho_csv = "../dados/tratados/PNAE_2024_limpo.csv"
df.to_csv(caminho_csv, index=False, encoding='utf-8-sig')

#uma versão em Excel, para facilitar a visualização dos dados, caso necessário.
caminho_excel = "../dados/tratados/PNAE_2024_limpo_referencia.xlsx"
df.to_excel(caminho_excel, index=False, engine='openpyxl')

print("Limpeza concluída!")

# Vendo a tabela final
df.head()
```
